```python
import numpy as np
import pandas as pd
#Create dataframe

data={'company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
     'Person':['Sam','charlie','Amy','vanessa','Carl','Sarah'], 
     'sales':[200,120,340,124,243,350]}

df=pd.DataFrame(data)
df
```

```python
df=pd.DataFrame(data)
df.groupby('company')
```

```python
by_comp=df.groupby("company")
print(by_comp)
```

```python
df.groupby('company')['sales'].mean()
```
